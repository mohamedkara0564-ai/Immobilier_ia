```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("logements_synthetiques.csv")
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df["prix"], bins=50, kde=True)
plt.title("Distribution des prix de location simulés")
plt.xlabel("Prix (FCFA)")
plt.show()
```

```python
prix_comm = df.groupby("quartier")["prix"].mean().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=prix_comm.values, y=prix_comm.index)
plt.title("Prix moyen par quartier")
plt.xlabel("Prix moyen (FCFA)")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="surface", y="prix", hue="type")
plt.title("Prix en fonction de la surface")
plt.show()
```
